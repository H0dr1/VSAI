```python

from sklearn.linear_model import LinearRegression
import numpy as np

X_train = np.array([[50], [80], [100], [120], [150]])


y_train = np.array([500, 800, 1000, 1200, 1500])


model = LinearRegression()
model.fit(X_train, y_train)

surface = int(input("Enter a house surface (m²): "))
new_house = np.array([[surface]])
predicted_price = model.predict(new_house)

print(f"prediction for a house have {surface} m² is: {predicted_price[0]:.2f}")
```
